## Factoriales

El factorial de un número natural es la multiplicación de todos los números anteriores hasta llegar a 1. Por ejemplo el factorial de 5 es igual a $5* 4* 3* 2* 1$, es decir, 120.

Igual que con la serie Fibonacci, hay dos formas de resolver este problema: una con un loop for y la otra utilizando funciones recursivas.  Recuerden que cada función recursiva tiene que tener una condición terminal.
En el caso de Fibonacci, la condición terminal es que fibo(1) = 1.

En el caso de factorial, la condición terminal será factorial(1) = 1

Escriban ahora las dos versiones posibles de la función factorial(x).

**SOLUCION RECURSIVA FIBONACCI**

Al pedir por ejemplo factorial(5) el bucle en return espera un resultado anterior con x-1 pero no existe en el unico lugar que existe un resultado es en factorial(1).

Desde aca inicia la cuenta fibonacci:

x*factorial(x-1)

2 * 1 = 2

3 * 2 = 6

4 * 6 = 24

5 * 24 = 120

x= factoriales que se quedaron experando resultados anteriores:

factorial(2) * (resultado anterior x-1)

factorial(3) * (resultado anterior x-1)

factorial(4) * (resultado anterior x-1)

factorial(5) * (resultado anterior x-1)


```python
#solucion recursiva
def factorial(x):
    if x == 1:
        return 1
    else:
        return x*factorial(x-1)
```

```python
factorial(5)

```

**SOLUCIÓN ITERATIVA FIBONACCI**

A diferencia del otro método este escala hacia delante.

factorial(5)

* resultado = 1

* range = rango entre (1, 5+1) es decir rango entre 1 y 6

* Luego solo la cuenta:

resultado = resultado * (i)

resultado = 1 * 1(primer rango) = 1

resultado = 1 * 2(segundo rango) = 2

resultado = 2 * 3(tercer rango) = 6

resultado = 6 * 4(cuarto rango) = 24

resultado = 24 * 5(quinto rango) = 120




```python
#Solucion recursiva
def factorial(x):
    resultado = 1
    for i in range(1, x+1):
        resultado = resultado * (i)
    return resultado
```

```python
factorial(5)
```

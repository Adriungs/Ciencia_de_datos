```python
def ruta(tupla):
    rutas = tupla[0]
    archivos = tupla [1]
    #print(f'rutas: {rutas}, archivos : {archivos}')
    with open(rutas, 'r') as visto:
      contenido = visto.read()
      #print(contenido)
      nueva_lista = contenido.split()
      #print(f'contenido_como_lista: {nueva_lista}')
      palabras = len(nueva_lista)
      print(f'{archivos} = {palabras}')
for tupla in tuplas_archivos:
    ruta(tupla)
```

##Devolver resultado

**Modificar la función para que:**

* devuelva una tupla (nombre_archivo, cantidad_palabras)
* NO usar print dentro de la función

```python
def ruta(tupla):
    rutas = tupla[0]
    archivos = tupla [1]
    #print(f'rutas: {rutas}, archivos : {archivos}')
    with open(rutas, 'r') as visto:
      contenido = visto.read()
      #print(contenido)
      nueva_lista = contenido.split()
      #print(f'contenido_como_lista: {nueva_lista}')
      palabras = len(nueva_lista)
      #print(f'{archivos} = {palabras}')
      return archivos, palabras
for tupla in tuplas_archivos:
    ruta(tupla)
for tupla in tuplas_archivos:
    archivo, cant = ruta(tupla)
    print(f'{archivo} = {cant}')
```

##Filtrar reseñas largas

**Usando el resultado de la función:**

mostrar qué archivos tienen más de 180 palabras

```python
def ruta(tupla):
    rutas = tupla[0]
    archivos = tupla [1]
    #print(f'rutas: {rutas}, archivos : {archivos}')
    with open(rutas, 'r') as visto:
      contenido = visto.read()
      #print(contenido)
      nueva_lista = contenido.split()
      #print(f'contenido_como_lista: {nueva_lista}')
      palabras = len(nueva_lista)
      #print(f'{archivos} = {palabras}')
      return archivos, palabras
for tupla in tuplas_archivos:
    archivo, cant = ruta(tupla)
    #print(f"Nombre: {archivo}, Palabras: {cant}")
    if cant > 50:
      print(f'este archivo: {archivo}, tiene {cant} palabras')
```
